```python
import os


def replace_mask_with_image(folder_path):
    for filename in os.listdir(folder_path):
        if 'label' in filename:
            new_filename = filename.replace('label', 'image')
            original = os.path.join(folder_path, filename)
            result = os.path.join(folder_path, new_filename)
            os.rename(original, result)

#need the same names for the next part
replace_mask_with_image('data/preprocesd/MMWHS/heart_masks')
```
